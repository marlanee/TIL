## 모든 위치

> 주어진 문자열(text)에서 제시된 알파벳(alphabet)의 등장 위치를 리스트로 반환하시오.
> 해당 알파벳이 등장하지 않으면, -1을 반환하시오.

---

예시)

```python
print(my_find('apple', 'p')) # [1, 2]
print(my_find('a', 'p'))     # -1
```


````python
def my_find(text, alphabet):
    find_str = []
    for my_str, small_str in enumerate(text):
        if small_str == alphabet:
            find_str.append(my_str)
    if not find_str:
        return -1
    return find_str
````

````python
# 해당 코드를 통해 올바른 결과가 나오는지 확인하시오.
print(my_find('apple', 'p'))
print(my_find('a', 'p'))
````

## 출석 체크

> 주어진 학생 n과 출석한 학생명부 students 문자열이 있다. 결석한 학생들로 구성된 문자열을 반환하시오.
>
> n이 7일 때, 1 2 3 4 5 6 7의 출석 번호가 부여되고,
>
> '1 3 5'는 출석한 학생 명부이다. 
>
> 즉, 결석한 학생 명부 '2 4 6 7'을 return 해야 한다.

---

예시)

```python
print(check(7, '1 3 5')) # 2 4 6 7
```


````python
def check(n, students):
    total_list = []
    for i in range(1, n + 1):
        total_list.append(i)

    total_set = set(total_list)
    no_student = set(map(int, students.split()))
    result = list((total_set - no_student))   # 세트 괄호를 어떻게 벗기지? -> 리스트로 만든 다음 join 메서드로 벗기자 / 숫자는 join 사용이 불가하다. 명심해라.
    
    return ' '.join(map(str, result))   # str로 형변환 후 map으로 자판기 생성 -> 자판기에서 join으로 뽑은 후 묶어서 반환
````

````python
# 해당 코드를 통해 올바른 결과가 나오는지 확인하시오.
print(check(7, '1 3 5')) # 2 4 6 7
````
