## 최댓값 구하기

> 주어진 리스트의 요소 중에서 최댓값을 출력하시오.

---
```
[출력 예시]
22
```

````python
numbers = [7, 10, 22, 4, 3, 17]

numbers.sort()
print(numbers[-1])   # 정렬 후 마지막 숫자 출력 / 하드코딩을 피했음
print(max(numbers))   # max 함수 쓸 경우

````

## 'a'가 싫어


> 입력으로 짧은 영단어 word가 주어질 때, 해당 단어에서 'a'를 모두 제거한 결과를 출력하시오.

---
```
[입력 예시]
apple

[출력 예시]
pple
```

````python
word = input()

word_list = []   # 메서드가 생각나지 않는다. 리스트로 돌려야겠다.
for i in word:
    if i != 'a':
        word_list.append(i)
print("".join(word_list))

# 세상에, 훨씬 쉬운 방법이 있었다. / replace 문자열 메서드로 a를 공백으로 대체
result = word.replace('a', '')
print(result)
````

## 단어 뒤집기

> 입력으로 짧은 영어단어 word가 주어질 때, 해당 단어를 역순으로 뒤집은 결과를 출력하시오.

---
```
[입력 예시]
apple

[출력 예시]
elppa
```

````python
word = input()

print(word[::-1])   # 슬라이싱을 쓰고 싶지는 않았는데.

print(''.join(reversed(word)))   # reversed 함수와 join 메서드 활용
````
